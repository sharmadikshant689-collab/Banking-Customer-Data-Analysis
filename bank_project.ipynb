```python
# Q8. Account Type Analysis
# For each Account_Type, find the number of customers, average account balance, average income, and average monthly transactions.
Account_Type_Analysis=df.groupby("Account_Type").agg(
    number_of_customers=("Gender","count"),
    average_account_balance=("Account_Balance","mean"),
    average_income=("Annual_Income","mean"),
    average_monthly_transactions=("Transactions_Per_Month","mean")
)
print(Account_Type_Analysis)
```

```python
# Q9. Credit Score Classification
# Create these categories based on Credit_Score:

# Excellent → 750+
# Good → 700–749
# Average → 650–699
# Poor → Below 650

# Then find the number of customers in each category.


df["Credit_Category"] = np.select(
    [
        df["Credit_Score"] >= 750,
        df["Credit_Score"] >= 700,
        df["Credit_Score"] >= 650
    ],
    [
        "Excellent",
        "Good",
        "Average"
    ],
    default="Poor"
)

print(df[["Credit_Score", "Credit_Category"]])
```

```python
 # Q10. Loan Analysis
# Find:

# Total number of customers having a loan
# Total loan amount
# Average loan amount
# Median loan amount
# Average interest rate
# Most common loan type
Loan_Analysis =df.groupby("Loan_Type").agg(
    number_customer_loan=("Loan_Type","count"),
    total_loan_amount=("Loan_Amount","sum"),
    avg_loan_amount=("Loan_Amount","mean"),
    median_loan_amount=("Loan_Amount","median"),
    avg_interest_rate=("Interest_Rate","mean")
)
print(Loan_Analysis)
```

```python
# Q11. High-Risk Customers
# Find customers who satisfy at least two of these conditions:

# Credit score < 650
# Loan amount > ₹10,00,000
# Interest rate > 15%
# Account balance < ₹20,000

# Display their Customer_ID, Customer_Name, Credit_Score, Loan_Amount, Interest_Rate, and Account_Balance.
# Q11. High-Risk Customers

risk_count = (
    (df["Credit_Score"] < 650).astype(int)
    + (df["Loan_Amount"] > 1000000).astype(int)
    + (df["Interest_Rate"] > 15).astype(int)
    + (df["Account_Balance"] < 20000).astype(int)
)

df["risk"] = np.select(
    [
        risk_count >= 3,
        risk_count == 2,
        risk_count == 1
    ],
    [
        "High Risk",
        "Medium Risk",
        "Low Risk"
    ],
    default="Safe"
)

risk1 = df[
    df["risk"].isin(["High Risk", "Medium Risk"])
][[
    "Customer_ID",
    "Customer_Name",
    "Credit_Score",
    "Loan_Amount",
    "Interest_Rate",
    "Account_Balance"
]]

risk1
```

```python

# Q16. Customer Segmentation
# Divide customers into four groups based on income and account balance:

# High Income + High Balance
# High Income + Low Balance
# Low Income + High Balance
# Low Income + Low Balance

# Compare the number of customers, average credit score, average transactions, and churn rate for each segment.

```

```python
# Q16. Customer Segmentation
# Divide customers into four groups based on income and account balance:

# High Income + High Balance
# High Income + Low Balance
# Low Income + High Balance
# Low Income + Low Balance

# Find median values
income_median = df["Annual_Income"].median()
balance_median = df["Account_Balance"].median()

# Create customer segments
df["Customer_Segment"] = np.where(
    (df["Annual_Income"] >= income_median) & 
    (df["Account_Balance"] >= balance_median),
    "High Income + High Balance",

    np.where(
        (df["Annual_Income"] >= income_median) & 
        (df["Account_Balance"] < balance_median),
        "High Income + Low Balance",

        np.where(
            (df["Annual_Income"] < income_median) & 
            (df["Account_Balance"] >= balance_median),
            "Low Income + High Balance",
            "Low Income + Low Balance"
        )
    )
)

# See the result
df[["Annual_Income", "Account_Balance", "Customer_Segment"]].head(10)
```
